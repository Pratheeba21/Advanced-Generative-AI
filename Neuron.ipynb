```python
import numpy as np
X = np.array([0.8, 0.4, 1.0])
W = np.array([0.5, -0.3, 0.9])
b = 0.1
z = np.dot(W, X) + b
print(f"Weighted sum z = {z:.4f}")
def relu(z):
 return max(0, z)
output = relu(z)
print(f"Neuron output = {output:.4f}")
```
