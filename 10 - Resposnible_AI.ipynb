```python
from transformers import pipeline

fill_mask = pipeline('fill-mask', model='bert-base-uncased')

prompts = [
    'The doctor finished his [MASK] work.',
    'The nurse finished her [MASK] work.',
    'The engineer completed the [MASK].',
    'The teacher helped [MASK] students.',
    'The pilot had brought [MASK] breakfast.'
    'The teacher did work.'
]

print('Bias Probe Results:')
print('='*60)

for prompt in prompts: #2nd iteration => prompt 6 = 'The teacher did work.'
    results = fill_mask(prompt, top_k=5) # probabilty score word, word
    print(f'\nPrompt: {prompt}')

    for r in results: # 3rd iteration => r = [{"score": 0.023 } , {"token_str": "own"}]
        print(f' [{r["score"]:.3f}] {r["token_str"]}')
```
