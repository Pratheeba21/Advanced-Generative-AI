```python
from transformers import pipeline #model, tokenizer, processing

classifier = pipeline('sentiment-analysis') # default - DistilBERT model
print(classifier('I absolutely love this generative AI course!'))

generator = pipeline('text-generation', model='gpt2')
output = generator('Generative AI is', max_new_tokens=50, num_return_sequences=3) # list = [{token_id, generated_text, attention_we},{},{}]
print(output[0]['generated_text'])
print(output[1]['generated_text'])
print(output[2]['generated_text'])
```
