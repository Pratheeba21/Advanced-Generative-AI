```python
from huggingface_hub import HfApi # HTTP req to HF BE DB

api = HfApi() #instantiation / object creation
models = list(
    api.list_models(
        pipeline_tag='text-generation',
        sort='downloads',
        limit=10
    )
)
#models = [m1, m2, m3, m4, m5]

print('Top 5 Text Generation Models:')
for m in models: # 2nd iteration => m = m2 (m2 => modelId, downloads)
    downloads = getattr(m, 'downloads', 0) or 0
    print(f' {m.modelId} — {downloads:,} downloads')

model_info = api.model_info('ibm-granite/granite-3.1-2b-instruct')
print('\nModel Tags:', model_info.tags)
```
