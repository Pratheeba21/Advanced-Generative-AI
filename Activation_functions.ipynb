```python
import numpy as np

def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
 return 1 / (1 + np.exp(-z))

def tanh_fn(z):
 return np.tanh(z)

test_values = [-3, -1, 0, 1, 3]
print(f"{'z':>6} | {'ReLU':>8} | {'Sigmoid':>8} | {'Tanh':>8}")
print("-" * 40)

for z in test_values:
 print(f"{z:>6} | {relu(z):>8.4f} | {sigmoid(z):>8.4f} | {tanh_fn(z):>8.4f}")
```
