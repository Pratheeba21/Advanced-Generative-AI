```python
from huggingface_hub import InferenceClient

client = InferenceClient(token='your_token')

print('\n--- Image Generation (via InferenceClient) ---')
try:
  image = client.text_to_image( #call the task directly
      model="stabilityai/stable-diffusion-xl-base-1.0",
      prompt='A photorealistic robot standing infront of a classroom teaching the students'
  )

  if image:
    image.save('generated_image_client5.jpg')
    print('Image saved as generated_image_client5.jpg!')
  else:
    print('Image generation returned no image.')
except Exception as e:
  print(f"Error during image generation with InferenceClient: {e}")
```
